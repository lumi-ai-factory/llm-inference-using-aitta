## Course introduction - what is this course about

In this concise online course, you'll learn the basics of using LLMs and their application through LUMI AI Factory's (LUMI AIF) inference service, Aitta. This course includes both theoretical material and interactive coding exercises. We will use models available in Aitta through its OpenAI-compatible API with the Python library [`openai`](https://pypi.org/project/openai/). Aitta offers generative large language models (LLMs), including models that accept image input, as well as text embedding models.

> Aitta is intended for research and development. It runs on the LUMI supercomputer, so there is no guarantee that a particular model is available at any given time, and the first request to a model can take a while because the model has to be started first. Capacity is shared between all users, so delays are possible especially when many users are active.

We'll provide a brief overview of model training and the basics of using them (inference) across different setups. **The primary focus will be on exploring Aitta and harnessing GPU power for LLM inference.**

This course is specifically centered on using LLMs with textual data. Other AI modalities and model types are not covered. While we touch on model training in the material, there are no hands-on exercises dedicated to this topic. 


## Course setup and prequisites - what do you need to complete this course

Some prior knowledge of machine learning and natural language processing will be beneficial. During the theoretical parts, you will be guided to do related coding exercises. Basic knowledge of Python programming and Jupyter notebooks is required for the exercises.

Coding exercises are available through [CSC's Noppe service](https://noppe.rahtiapp.fi/welcome) for interactive web based applications ([CSC Docs](https://docs.csc.fi/cloud/noppe/)). A dedicated workspace is set up for this course, allowing you to utilize an inference service [Aitta](https://aitta.csc.fi/) within a Jupyter notebook environment.  For more information on using Jupyter Notebooks, visit [Jupyter Documentation](https://docs.jupyter.org/en/latest/index.html). 

To access Noppe, you can log in with a CSC account, Haka, Virtu or MOOC.fi credentials. To use Aitta, you need a valid LUMI account with access to at least one active LUMI project. You can log in to Aitta with your Haka, MyAccessId or CSC account.

If you prefer to complete Jupyter notebook exercises in your own environment, ensure you have Python 3.11 or higher, along with the libraries listed in the [requirements.txt](../exercises/requirements.txt) file.
